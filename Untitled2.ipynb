```python
import matplotlib.pyplot as plt

# Define timeline segments with labels and durations in minutes
segments = [
    ("Spontaneous", 60), 
    ("Light Stimulus", 5),  # This will be split into subsegments
    (" ", 10), 
    ("Drug Application", 5), 
    ("Drug", 60)
]

# Convert minutes to cumulative positions (in seconds)
start_times = [0]
for _, duration in segments:
    start_times.append(start_times[-1] + duration * 60)  # Convert minutes to seconds

# Create figure
fig, ax = plt.subplots(figsize=(20, 2))
ax.set_xlim(0, start_times[-1])  # Set x-axis limit

# Colors for different phases
colors = {
    "Spontaneous": "lightgrey",
    " ": "lightgrey",
    "Drug Application": "lightgrey",
    "Drug": "lightgrey",
}

# Plot timeline segments except Light Dark Cycle (which will be split)
for i, (label, duration) in enumerate(segments):
    if label != "Light Stimulus":
        ax.broken_barh([(start_times[i], duration * 60)], (0, 1), facecolors=colors[label], edgecolor="none", alpha=0.9)

# Light-Dark Cycle (split into 10 alternating subsegments)
ld_start = start_times[1]  # Start time of Light-Dark cycle
ld_durations = [(10, "lightgrey"), (50, "black")]  # 10 sec light, 50 sec dark
num_repeats = 5

for i in range(num_repeats):
    for duration, color in ld_durations:
        ax.broken_barh([(ld_start, duration)], (0, 1), facecolors=color, edgecolor="none", alpha=0.9)
        ld_start += duration

# Format x-axis: Only set ticks at segment starts
ax.set_xticks(start_times)
ax.set_xticklabels([f"{t//60} min" for t in start_times], rotation=45, fontsize=10, color="black")


# Make x-axis, ticks, and labels black
ax.spines['bottom'].set_color("black")  # X-axis line
ax.grid(color='black', linestyle='--', linewidth=1)

# Remove y-axis ticks and unwanted spines
ax.set_yticks([])
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)


# Move labels below the x-axis
for i, (label, duration) in enumerate(segments):
    ax.text(start_times[i] + (duration * 30), 1.2, label, ha='center', va='center_baseline', fontsize=10, fontweight='bold')

# Title
plt.title("Behavioural Paradigm", fontsize=12, fontweight="bold", pad=30)

# Save figure with transparent background
plt.savefig("behavioural_paradigm.png", dpi=300, bbox_inches='tight', transparent=True)

plt.show()
```
